```python
def edit_distance(word1, word2):
    m = len(word1)
    n = len(word2)

    dp = [[0 for _ in range(n + 1)] for _ in range(m + 1)]

    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if word1[i - 1] == word2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                dp[i][j] = 1 + min(
                    dp[i - 1][j],
                    dp[i][j - 1],
                    dp[i - 1][j - 1]
                )

    return dp[m][n]


word1 = "intention"
word2 = "execution"

print("Edit Distance:", edit_distance(word1, word2))
```

```python
def edit_distance_with_operations(word1, word2):
    m = len(word1)
    n = len(word2)

    dp = [[0 for _ in range(n + 1)] for _ in range(m + 1)]

    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if word1[i - 1] == word2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                dp[i][j] = 1 + min(
                    dp[i - 1][j],
                    dp[i][j - 1],
                    dp[i - 1][j - 1]
                )

    i, j = m, n
    operations = []

    while i > 0 or j > 0:
        if i > 0 and j > 0 and word1[i - 1] == word2[j - 1]:
            i -= 1
            j -= 1
        elif i > 0 and dp[i][j] == dp[i - 1][j] + 1:
            operations.append(f"Delete '{word1[i-1]}'")
            i -= 1
        elif j > 0 and dp[i][j] == dp[i][j - 1] + 1:
            operations.append(f"Insert '{word2[j-1]}'")
            j -= 1
        else:
            operations.append(f"Substitute '{word1[i-1]}' with '{word2[j-1]}'")
            i -= 1
            j -= 1

    operations.reverse()

    print("Edit Distance:", dp[m][n])
    print("Operations:")
    for op in operations:
        print(op)


word1 = "execution"
word2 = "intuition"

edit_distance_with_operations(word1, word2)
```

```python
def edit_distance_with_costs(word1, word2, insertion_cost=1, deletion_cost=1, substitution_cost=1):
    m = len(word1)
    n = len(word2)

    dp = [[0 for _ in range(n + 1)] for _ in range(m + 1)]

    for i in range(m + 1):
        dp[i][0] = i * deletion_cost
    for j in range(n + 1):
        dp[0][j] = j * insertion_cost

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if word1[i - 1] == word2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                dp[i][j] = min(
                    dp[i - 1][j] + deletion_cost,
                    dp[i][j - 1] + insertion_cost,
                    dp[i - 1][j - 1] + substitution_cost
                )

    return dp[m][n]

def edit_distance(word1, word2):
    m = len(word1)
    n = len(word2)

    dp = [[0 for _ in range(n + 1)] for _ in range(m + 1)]

    for i in range(m + 1):
        dp[i][0] = i
    for j in range(n + 1):
        dp[0][j] = j

    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if word1[i - 1] == word2[j - 1]:
                dp[i][j] = dp[i - 1][j - 1]
            else:
                dp[i][j] = 1 + min(
                    dp[i - 1][j],
                    dp[i][j - 1],
                    dp[i - 1][j - 1]
                )
    return dp[m][n]


word1 = "book"
word2 = "back"

insertion_cost = 1
del_cost = 1
sub_cost = 2

print(f"Word 1: '{word1}'")
print(f"Word 2: '{word2}'")
print("----------------------------------------")

ed_with_costs = edit_distance_with_costs(word1, word2, insertion_cost, del_cost, sub_cost)
print(f"Edit Distance with custom costs (Insert={insertion_cost}, Delete={del_cost}, Substitute={sub_cost}): {ed_with_costs}")

uniform_ed = edit_distance(word1, word2)
print(f"Uniform-cost Edit Distance: {uniform_ed}")

print("----------------------------------------")
print("Comparison: ")
if ed_with_costs == uniform_ed:
    print("The edit distances are the same.")
else:
    print(f"The edit distance with custom costs ({ed_with_costs}) is different from the uniform-cost edit distance ({uniform_ed}).")
```

```python
import torch
import numpy as np
```

```python
tensor_from_list = torch.tensor([1, 2, 3])
print("Tensor from list:", tensor_from_list)
```

```python
np_array = np.array([[1, 2], [3, 4]])
tensor_from_numpy = torch.from_numpy(np_array)
print("Tensor from NumPy array:\n", tensor_from_numpy)
```

```python
a = torch.tensor([1, 2, 3])
b = torch.tensor([1, 2, 3])

addition = a + b
print("Addition Result:", addition)

multiplication = a * a
print("Multiplication Result:", multiplication)
```

```python
print("Tensor:", a)
print("Tensor Shape:", a.shape)
print("Tensor Data Type:", a.dtype)
```

```python
A = torch.tensor([[1, 2],
                  [3, 4]])

B = torch.tensor([[5, 6],
                  [7, 8]])

AxB = torch.mm(A, B)

print("Matrix A:\n", A)
print("Matrix B:\n", B)
print("A x B:\n", AxB)
```

```python
x = torch.tensor([1, 3, 5])
y = torch.tensor([2, 4, 6])

dot_product = torch.dot(x, y)
print("Dot Product Output =", dot_product)
```

```python
T = torch.tensor([[10, 20, 30],
                  [40, 50, 60]])

print("Tensor T:\n", T)

print("T[0][1]:", T[0][1])

print("First row:", T[0])

print("Second column:", T[:, 1])
```

```python
print("torch.empty():\n", torch.empty(2, 3))

print("torch.zeros():\n", torch.zeros(2, 3))

print("torch.rand():\n", torch.rand(2, 3))

print("torch.ones():\n", torch.ones(2, 3))

print("torch.eye():\n", torch.eye(3))

print("torch.arange():", torch.arange(0, 10, 2))

print("torch.linspace():", torch.linspace(0, 1, steps=5))

x = torch.tensor([1, 2])
y = torch.tensor([3, 4])
print("torch.add():", torch.add(x, y))

print("torch.mm():\n", torch.mm(A, B))

print("torch.dot():", torch.dot(x, y))

batch1 = torch.rand(2, 3, 4)
batch2 = torch.rand(2, 4, 5)
print("torch.bmm(): shape", torch.bmm(batch1, batch2).shape)

print("torch.abs():", torch.abs(torch.tensor([-3, -1, 2])))

print("torch.argmax():", torch.argmax(torch.tensor([10, 50, 20])))

print("torch.argmin():", torch.argmin(torch.tensor([10, 50, 20])))

print("torch.mean():", torch.mean(torch.tensor([1.0, 2.0, 3.0])))

print("torch.sort():", torch.sort(torch.tensor([4, 1, 3, 2])))

print("torch.cat():", torch.cat((torch.tensor([1, 2]), torch.tensor([3, 4]))))
```

```python
import torch.nn as nn

model = nn.Linear(in_features=1, out_features=1)

x = torch.tensor([[1.0], [2.0], [3.0]])

y_pred = model(x)

print("Input:\n", x)
print("Predicted Output:\n", y_pred)
```

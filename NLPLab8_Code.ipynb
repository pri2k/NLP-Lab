```python
import kagglehub

dataset_path = kagglehub.dataset_download("jacksondivakarr/sample34")
print(f"Dataset downloaded to: {dataset_path}")
```

```python
import pandas as pd
import os

# List files in the downloaded dataset directory
downloaded_files = os.listdir(dataset_path)
print(f"Files found in the dataset directory: {downloaded_files}")

# Try to find a CSV file. Assuming there's only one relevant CSV file in the directory.
csv_file_name = None
for file in downloaded_files:
    if file.endswith('.csv'):
        csv_file_name = file
        break

if csv_file_name:
    csv_file_path = os.path.join(dataset_path, csv_file_name)
    print(f"Found CSV file: {csv_file_name}. Loading data from: {csv_file_path}")
    df = pd.read_csv(csv_file_path)

    # Display the first few rows of the DataFrame to verify
    print("Dataset loaded successfully. Displaying the first 5 rows:")
    print(df.head().to_markdown(index=False))
else:
    print("Error: No CSV file was found in the downloaded dataset directory.")
    df = None # Ensure df is defined even if no CSV is found
```

```python
print("First 5 rows of the DataFrame:")
print(df.head().to_markdown(index=False))

print("\nValue counts for 'fuel' column:")
print(df['fuel'].value_counts().to_markdown(numalign="left", stralign="left"))

print("\nValue counts for 'owner' column:")
print(df['owner'].value_counts().to_markdown(numalign="left", stralign="left"))

print("\nValue counts for 'name' column:")
print(df['name'].value_counts().to_markdown(numalign="left", stralign="left"))
```

```python
df_encoded_standard = pd.get_dummies(df[['fuel', 'owner']])
print("First 5 rows of the one-hot encoded DataFrame (df_encoded_standard):")
print(df_encoded_standard.head().to_markdown(index=False))
```

```python
df_encoded_k1 = pd.get_dummies(df[['fuel', 'owner']], drop_first=True)
print("First 5 rows of the one-hot encoded DataFrame (df_encoded_k1) with K-1 encoding:")
print(df_encoded_k1.head().to_markdown(index=False))
```

```python
from sklearn.preprocessing import OneHotEncoder

# Instantiate OneHotEncoder with handle_unknown='ignore' and sparse_output=False
ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

# Select the categorical columns to encode
categorical_cols = ['fuel', 'owner']
df_categorical = df[categorical_cols]

# Fit and transform the selected columns
df_encoded_array = ohe.fit_transform(df_categorical)

# Create a new DataFrame from the encoded array with feature names
df_encoded_ohe = pd.DataFrame(df_encoded_array, columns=ohe.get_feature_names_out(categorical_cols))

# Concatenate the new encoded DataFrame with the original df, dropping the original categorical columns
df_encoded_sklearn = pd.concat([df.drop(columns=categorical_cols), df_encoded_ohe], axis=1)

# Display the first 5 rows of df_encoded_sklearn to verify
print("First 5 rows of the DataFrame after One-Hot Encoding with Scikit-learn:")
print(df_encoded_sklearn.head().to_markdown(index=False))
```

```python
import pandas as pd

# 1. Calculate the value counts for the 'name' column
name_counts = df['name'].value_counts()

# 2. Define a threshold for category frequency
threshold = 100 # Example threshold

# 3. Identify the 'name' categories that have counts below this threshold
infrequent_names = name_counts[name_counts < threshold].index

# 4. Create a new column, df['name_encoded'], by copying the 'name' column
df_encoded_top_categories = df.copy()
df_encoded_top_categories['name_encoded'] = df_encoded_top_categories['name']

# 5. Replace the 'name' categories in df['name_encoded'] that are below the threshold with the string 'uncommon'
df_encoded_top_categories.loc[df_encoded_top_categories['name_encoded'].isin(infrequent_names), 'name_encoded'] = 'uncommon'

# 6. Apply pd.get_dummies() to the df['name_encoded'] column to one-hot encode it.
df_name_ohe = pd.get_dummies(df_encoded_top_categories['name_encoded'], prefix='name')

# 7. Concatenate the resulting one-hot encoded DataFrame with the original df,
# dropping the original 'name' column and the temporary 'name_encoded' column to avoid redundancy.
df_encoded_top_categories = pd.concat([
    df.drop(columns=['name']),
    df_name_ohe
], axis=1)

# 8. Print the first 5 rows of the df_encoded_top_categories DataFrame to verify the changes.
print("First 5 rows of the DataFrame after encoding top 'name' categories:")
print(df_encoded_top_categories.head().to_markdown(index=False))
```

# Assignment 2

```python
news_headlines = [
    'Congress set to approve new spending bill',
    'Stock market surges on positive economic news',
    'Local elections see record voter turnout',
    'New technology promises to revolutionize healthcare',
    'Climate change remains a global concern'
]

preprocessed_headlines = []

for headline in news_headlines:
    # Convert to lowercase and split into words
    tokens = headline.lower().split()
    preprocessed_headlines.append(tokens)

print("Preprocessed Headlines:")
print(preprocessed_headlines)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

# Instantiate CountVectorizer with default parameters
vectorizer = CountVectorizer()

# Fit the vectorizer to the original news_headlines list and transform the text data into a Term-Document Matrix
tdm = vectorizer.fit_transform(news_headlines)

# Convert the tdm into a dense NumPy array for easier viewing
tdm_dense = tdm.toarray()

# Get the feature names (vocabulary) from the fitted CountVectorizer
vocabulary = vectorizer.get_feature_names_out()

print("Shape of the Term-Document Matrix (TDM):", tdm_dense.shape)
print("\nFirst 5 rows of the dense TDM:")
print(tdm_dense[:5])
print("\nVocabulary (feature names):", vocabulary)
```

```python
print("Document Embeddings (tdm_dense):")
print(tdm_dense)

print("\nBasis Vectors (Terms in Vocabulary):")
print(vocabulary)
```

```python
print("Word Embeddings (Transpose of TDM):")
print(tdm_dense.T)

print("\nBasis Vectors (Documents/News Headlines):")
print(news_headlines)
```

```python
print("\n--- Example from tdm_dense ---")

# Let's pick the first document (index 0) as an example
document_index = 0
example_document_tf = tdm_dense[document_index]
example_headline = news_headlines[document_index]

print(f"\nDocument: \"{example_headline}\"")
print(f"TF Vector for this document: {example_document_tf}")

# Get the terms that appear in this document (TF > 0)
terms_in_document = []
for i, tf_value in enumerate(example_document_tf):
    if tf_value > 0:
        terms_in_document.append((vocabulary[i], tf_value))

print("\nTerms and their frequencies in this document:")
for term, freq in terms_in_document:
    print(f"- {term}: {freq}")

print("\nIn this example, words like 'approve', 'bill', 'congress', 'new', 'set', 'spending', 'to' are present. Their non-zero counts in the TF vector indicate their relevance to this specific headline.")
```

```python
from sklearn.datasets import fetch_20newsgroups

# 1. Load the 20 Newsgroups dataset
# By default, fetch_20newsgroups downloads the data if not already present.
newsgroups_data = fetch_20newsgroups(subset='all', remove=('headers', 'footers', 'quotes'))

# 2. Access the news articles
all_news_text = newsgroups_data.data

# 3. Print the total number of documents loaded
print(f"Total number of documents loaded: {len(all_news_text)}")

# 4. Print the first 500 characters of the first document
if all_news_text:
    print(f"\nFirst 500 characters of the first document:\n{all_news_text[0][:500]}")
else:
    print("No documents were loaded.")
```

```python
import nltk
import string

# Download the 'punkt' tokenizer data if not already present
try:
    nltk.data.find('tokenizers/punkt')
    print("'punkt' tokenizer already downloaded.")
except LookupError:
    nltk.download('punkt')
    print("'punkt' tokenizer downloaded.")

# Download the 'punkt_tab' tokenizer data if not already present (required by some NLTK versions/functions)
try:
    nltk.data.find('tokenizers/punkt_tab')
    print("'punkt_tab' tokenizer already downloaded.")
except LookupError:
    nltk.download('punkt_tab')
    print("'punkt_tab' tokenizer downloaded.")

# Download the 'stopwords' corpus if not already present
try:
    nltk.data.find('corpora/stopwords')
    print("'stopwords' corpus already downloaded.")
except LookupError:
    nltk.download('stopwords')
    print("'stopwords' corpus downloaded.")
```

```python
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import string

# Get English stopwords
stop_words = set(stopwords.words('english'))

# Define the preprocessing function
def preprocess_text(text):
    # Convert to lowercase
    text = text.lower()

    # Tokenize the text
    tokens = word_tokenize(text)

    # Remove punctuation and non-alphanumeric tokens
    tokens = [word for word in tokens if word.isalpha()]

    # Remove stopwords
    tokens = [word for word in tokens if word not in stop_words]

    return tokens

# Apply the preprocessing function to each document
preprocessed_documents = [preprocess_text(doc) for doc in all_news_text]

# Print the first preprocessed document to verify the output
print(f"Original first document:\n{all_news_text[0][:200]}...")
print(f"\nFirst preprocessed document:\n{preprocessed_documents[0]}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

# Convert list of token lists back to list of strings
processed_texts = [" ".join(doc) for doc in preprocessed_documents]

# Instantiate CountVectorizer and create TF matrix
vectorizer_tf = CountVectorizer()
tf_matrix = vectorizer_tf.fit_transform(processed_texts)

# Instantiate TfidfVectorizer and create TF-IDF matrix
vectorizer_tfidf = TfidfVectorizer()
tfidf_matrix = vectorizer_tfidf.fit_transform(processed_texts)

# Print the shape of the matrices
print(f"Shape of TF Matrix: {tf_matrix.shape}")
print(f"Shape of TF-IDF Matrix: {tfidf_matrix.shape}")
```

```python
import numpy as np

# 1. Convert tf_matrix and tfidf_matrix to dense arrays
tf_dense = tf_matrix.toarray()
tfidf_dense = tfidf_matrix.toarray()

# 2. Calculate the mean TF score for each word
average_tf = np.mean(tf_dense, axis=0)

# 3. Calculate the mean TF-IDF score for each word
average_tfidf = np.mean(tfidf_dense, axis=0)

# 4. Retrieve the vocabulary
vocabulary = vectorizer_tf.get_feature_names_out()

# 5. Create a Pandas DataFrame
comparison_df = pd.DataFrame({
    'Word': vocabulary,
    'Average TF': average_tf,
    'Average TF-IDF': average_tfidf
})

# 6. Sort the DataFrame by 'Average TF' in descending order and display the head
print("\nTop words by Average TF:\n")
print(comparison_df.sort_values(by='Average TF', ascending=False).head(10).to_markdown(index=False))

# 7. Sort the DataFrame by 'Average TF-IDF' in descending order and display the head
print("\nTop words by Average TF-IDF:\n")
print(comparison_df.sort_values(by='Average TF-IDF', ascending=False).head(10).to_markdown(index=False))
```

```python
import requests
import re
import random
from collections import Counter
from nltk.util import ngrams

# -----------------------------
# Step 1: Load Hamlet Corpus
# -----------------------------
url = "https://raw.githubusercontent.com/teropa/nlp/master/resources/corpora/gutenberg/shakespeare-hamlet.txt"
text = requests.get(url).text

# -----------------------------
# Step 2: Preprocess Text
# -----------------------------
text = text.lower()
text = re.sub(r'[^a-z\s]', '', text)  # remove punctuation
tokens = text.split()

print("Total words in corpus:", len(tokens))

# -----------------------------
# Step 3: Generate Unigrams
# -----------------------------
unigrams = Counter(tokens)

print("\nTop 10 Unigrams:")
for word, freq in unigrams.most_common(10):
    print(word, ":", freq)

# -----------------------------
# Step 4: Generate Bigrams
# -----------------------------
bigram_list = list(ngrams(tokens, 2))
bigrams = Counter(bigram_list)

print("\nTop 10 Bigrams:")
for pair, freq in bigrams.most_common(10):
    print(pair, ":", freq)

# -----------------------------
# Step 5: Generate Trigrams
# -----------------------------
trigram_list = list(ngrams(tokens, 3))
trigrams = Counter(trigram_list)

# -----------------------------
# Step 6: Select 10 Random Unique Words
# -----------------------------
unique_words = list(set(tokens))
random_words = random.sample(unique_words, 10)

print("\nRandom 10 words:")
print(random_words)

# -----------------------------
# Step 7: Unigram Language Model
# -----------------------------
total_words = len(tokens)

print("\nUnigram Probabilities:")
for word in random_words:
    prob = unigrams[word] / total_words
    print(f"P({word}) = {prob}")

# -----------------------------
# Step 8: Bigram Language Model
# -----------------------------
print("\nBigram Probabilities:")
for bigram in bigram_list[:10]:
    w1, w2 = bigram
    prob = bigrams[(w1, w2)] / unigrams[w1]
    print(f"P({w2}|{w1}) = {prob}")

# -----------------------------
# Step 9: Trigram Language Model
# -----------------------------
print("\nTrigram Probabilities:")
for trigram in trigram_list[:10]:
    w1, w2, w3 = trigram
    bigram_count = bigrams[(w1, w2)]
    if bigram_count != 0:
        prob = trigrams[(w1, w2, w3)] / bigram_count
        print(f"P({w3}|{w1},{w2}) = {prob}")
```

```python
import requests
import re
import random
from collections import Counter
from nltk.util import ngrams
import math

# ----------------------------
# Load Hamlet Corpus
# ----------------------------
url = "https://raw.githubusercontent.com/teropa/nlp/master/resources/corpora/gutenberg/shakespeare-hamlet.txt"
text = requests.get(url).text

# preprocessing
text = text.lower()
text = re.sub(r'[^a-z\s]', '', text)
tokens = text.split()

print("Total tokens:", len(tokens))

# ----------------------------
# Assignment 1
# Unigram + Bigram frequency
# ----------------------------

unigrams = Counter(tokens)

bigram_list = list(ngrams(tokens,2))
bigrams = Counter(bigram_list)

print("\nTop 10 Unigrams")
for w,c in unigrams.most_common(10):
    print(w,c)

print("\nTop 10 Bigrams")
for b,c in bigrams.most_common(10):
    print(b,c)

# ----------------------------
# Assignment 2
# Sentence probability
# ----------------------------

trigram_list = list(ngrams(tokens,3))
trigrams = Counter(trigram_list)

total_words = len(tokens)

def unigram_sentence_prob(sentence):
    words = sentence.split()
    prob = 1
    for w in words:
        prob *= unigrams[w] / total_words
    return prob


def bigram_sentence_prob(sentence):
    words = sentence.split()
    prob = 1
    for i in range(1,len(words)):
        prob *= bigrams[(words[i-1],words[i])] / unigrams[words[i-1]]
    return prob


def trigram_sentence_prob(sentence):
    words = sentence.split()
    prob = 1
    for i in range(2,len(words)):
        prob *= trigrams[(words[i-2],words[i-1],words[i])] / bigrams[(words[i-2],words[i-1])]
    return prob


sentence = "to be or not to be"

print("\nSentence:", sentence)
print("Unigram Probability:", unigram_sentence_prob(sentence))
print("Bigram Probability:", bigram_sentence_prob(sentence))
print("Trigram Probability:", trigram_sentence_prob(sentence))

# ----------------------------
# Assignment 3
# LM for 10 random words
# ----------------------------

unique_words = list(set(tokens))
random_words = random.sample(unique_words,10)

print("\nRandom words:",random_words)

print("\nUnigram LM")
for w in random_words:
    print(w, unigrams[w]/total_words)

print("\nBigram LM examples")
for bg in bigram_list[:10]:
    w1,w2 = bg
    print(f"P({w2}|{w1}) =",bigrams[(w1,w2)]/unigrams[w1])

print("\nTrigram LM examples")
for tg in trigram_list[:10]:
    w1,w2,w3 = tg
    if bigrams[(w1,w2)] != 0:
        print(f"P({w3}|{w1},{w2}) =", trigrams[(w1,w2,w3)]/bigrams[(w1,w2)])

# ----------------------------
# Assignment 4
# Next word prediction
# ----------------------------

def predict_next_word(word):
    candidates = {}

    for (w1,w2),count in bigrams.items():
        if w1 == word:
            candidates[w2] = count

    if not candidates:
        return None

    return max(candidates,key=candidates.get)

print("\nNext word after 'the':",predict_next_word("the"))

# ----------------------------
# Rare Words
# ----------------------------

rare_words = [w for w,c in unigrams.items() if c==1]

print("\nFirst 20 Rare Words")
print(rare_words[:20])

# ----------------------------
# Perplexity
# ----------------------------

def perplexity_unigram(sentence):

    words = sentence.split()
    prob = unigram_sentence_prob(sentence)

    N = len(words)

    return pow(1/prob,1/N)


def perplexity_bigram(sentence):

    words = sentence.split()
    prob = bigram_sentence_prob(sentence)

    N = len(words)

    return pow(1/prob,1/N)


def perplexity_trigram(sentence):

    words = sentence.split()
    prob = trigram_sentence_prob(sentence)

    N = len(words)

    return pow(1/prob,1/N)


print("\nPerplexity")
print("Unigram:",perplexity_unigram(sentence))
print("Bigram:",perplexity_bigram(sentence))
print("Trigram:",perplexity_trigram(sentence))
```
